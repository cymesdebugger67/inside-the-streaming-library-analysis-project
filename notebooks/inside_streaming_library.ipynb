Historical four-platform movie catalogue analysis. This notebook reads validated stage outputs; it does not refresh the dataset or overwrite results. Platform flags overlap, and release year is not catalogue-entry year.

```python
from pathlib import Path
import csv
import json

root = Path.cwd()
if not (root / 'outputs/phase04_comparative/comparative_qa.json').is_file():
    root = root.parent
assert (root / 'outputs/phase04_comparative/comparative_qa.json').is_file(), 'Run Phase 01–04 from the project root first'

def table(relative):
    with (root / relative).open(encoding='utf-8-sig', newline='') as handle:
        return list(csv.DictReader(handle))

def manifest(relative):
    return json.loads((root / relative).read_text(encoding='utf-8'))

source = manifest('outputs/source_viability/source_audit.json')
prep = manifest('outputs/phase02_preparation/preparation_qa.json')
eda = manifest('outputs/phase03_eda/eda_qa.json')
comparative = manifest('outputs/phase04_comparative/comparative_qa.json')
print('Source movies:', source['qa']['source_rows'])
print('Platform memberships:', prep['platform_membership_rows'])
print('Genre assignments:', prep['genre_assignment_rows'])
print('Phase 04 tables:', comparative['table_rows'])
```

## Catalogue sizes and overlap

Platform membership counts can sum to more than the source movie population. A pairwise Jaccard index measures intersection divided by union, not a share of current streaming availability.

```python
summary = table('outputs/phase03_eda/platform_summary.csv')
overlap = table('outputs/phase03_eda/overlap.csv')
for row in summary: print(row)
print('Pairwise comparisons:', len(overlap))
for row in overlap: print(row)
```

## Interpretation boundary
