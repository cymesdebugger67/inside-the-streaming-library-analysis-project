Historical descriptive movie catalogue comparison; this is an undated snapshot. Release year is not platform listing date. No current availability, viewership, or market-share claims.


```python
from pathlib import Path
import sys
root=Path.cwd()
if not (root/"outputs/phase05_metadata_reviewed/metadata_qa.json").exists():
    root=root.parent
assert (root/"outputs/phase05_metadata_reviewed/metadata_qa.json").exists(), "Run corrected Phase 05 first"
sys.path.insert(0, str(root))
from streaming_library.integrated import load_project, filter_ids, describe_selection
movies,memberships,genres,metadata=load_project(root)
all_ids=filter_ids(movies,memberships,genres,metadata)
print("Verified source movie records:", len(all_ids))
```

## 1. Catalogue Grain and Platform Overlap

Unique source movies are not equivalent to the sum of non-exclusive platform memberships.


```python
full=describe_selection(all_ids,movies,memberships,genres)
print("Unique movie records:",full["unique_movie_records"])
print("Memberships:",full["platform_memberships"])
print("Multi-platform movies:",full["multi_platform_movies"])
for p,n in full["platform_movie_counts"].items(): print(p,n)
```

## 2. Release-year Composition

Cross-sectional historical composition is not a platform growth trend.


```python
eras=["before_1950","1950_1979","1980_1999","2000_2009","2010_plus"]
for e in eras:
    ids=filter_ids(movies,memberships,genres,metadata,era=e)
    print(e,len(ids))
```

## 3. Genre Representations

Movies can have multiple genres. These counts are not mutually exclusive.


```python
for row in full["top_genres"][:12]: print(row)
```

## 4. Observed Ratings and Runtime

Means condition on field availability and may be selection biased. Do not interpret as catalogue quality rankings.


```python
for key in ("imdb","rt","runtime"):
    print(key,"observed",full[f"{key}_observed_n"],"mean",full[f"{key}_mean_observed"])
```

## 5. Corrected Metadata Interpretation

Named-language counts exclude literal `None`; raw strings remain upstream. Country/language labels do not independently verify origins or audio availability.


```python
for dim in ("country","language"):
    available=sum(bool(metadata[dim][mid]) for mid in movies)
    print(dim,"movie records with >=1 named label", available,"of",len(movies))
```

## 6. Cross-Filter Consistency Check

The same underlying record can have several platforms and genres.


```python
selection=filter_ids(movies,memberships,genres,metadata,platform="Netflix",genre="Drama")
summary=describe_selection(selection,movies,memberships,genres)
print("Netflix-and-Drama movie records:",summary["unique_movie_records"])
print("All memberships among selected movies:",summary["platform_memberships"])
```
