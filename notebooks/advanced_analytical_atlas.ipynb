## Evidence trail

Eight tables and six visuals are derived from the verified source. Displayed results are not market shares, platform listing dates, or inferential estimates.

```python
from pathlib import Path
import csv, json, hashlib
from IPython.display import display, Image, Markdown
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
output = root / "outputs" / "phase08_advanced"
manifest = json.loads((output / "advanced_qa.json").read_text(encoding="utf-8"))
for name, expected in manifest["output_sha256"].items():
    actual = hashlib.sha256((output/name).read_bytes()).hexdigest()
    assert actual == expected, name
assert manifest["source_population"] == 16744
print("Verified", len(manifest["output_sha256"]), "outputs;", len(manifest["table_rows"]), "tables")
```

```python
def table(name, limit=12):
    with (output / (name + ".csv")).open(encoding="utf-8", newline="") as source:
        rows=list(csv.DictReader(source))
    print(name, "|", len(rows), "rows")
    display(rows[:limit])
    return rows
```
